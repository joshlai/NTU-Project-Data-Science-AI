```python
from google.cloud import bigquery
from pandas_datareader import data as pdr
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import datetime

# === CONFIG ===
project_id = "module2project-462306"           # Your GCP project ID
dataset_id = "stock_analysis"       # Your dataset
table_security = "fact_security"   # Your table with price data

client = bigquery.Client(project=project_id)

# Define ETF tickers for sectors (11 ETFs)
etf_tickers = {
    "XLK": "Technology",
    "XLY": "Consumer Discretionary",
    "XLC": "Communication Services",
    "XLI": "Industrials",
    "XLB": "Materials",
    "XLE": "Energy",
    "XLP": "Consumer Staples",
    "XLV": "Health Care",
    "XLU": "Utilities",
    "XLF": "Financials",
    "XLRE": "Real Estate"
}

# Define FRED economic indicators (6 series)
fred_series = {
    "FEDFUNDS": "Fed Funds Rate",
    "CPIAUCSL": "CPI",
    "UNRATE": "Unemployment Rate",
    "INDPRO": "Industrial Production",
    "DGS10": "10Y Treasury Yield",
    "UMCSENT": "Consumer Sentiment"
}

# Date range: last 5 years from today
end_date = datetime.datetime.today()
start_date = end_date - pd.DateOffset(years=5)

print(f"Fetching data from {start_date.date()} to {end_date.date()}")

# === Step 1: Load daily ETF prices from BigQuery ===
query_etf = f"""
SELECT date, ticker_symbol, close_price
FROM `{project_id}.{dataset_id}.{table_security}`
WHERE ticker_symbol IN ({','.join(f"'{t}'" for t in etf_tickers.keys())})
  AND date BETWEEN '{start_date.date()}' AND '{end_date.date()}'
"""
df_etf = client.query(query_etf).result().to_dataframe()
df_etf["date"] = pd.to_datetime(df_etf["date"])

# Pivot to wide format: one column per ETF ticker
etf_prices = df_etf.pivot(index="date", columns="ticker_symbol", values="close_price")

# Resample monthly and then quarterly (last day of period)
etf_prices_m = etf_prices.resample("M").last()
etf_prices_q = etf_prices_m.resample("Q").last()

print("Quarterly ETF prices head:")
print(etf_prices_q.head())

# === Step 2: Download daily FRED economic data ===
econ_data = pd.DataFrame()
for series_id in fred_series.keys():
    df = pdr.DataReader(series_id, 'fred', start_date, end_date)
    monthly = df.resample('M').last()
    monthly.rename(columns={series_id: fred_series[series_id]}, inplace=True)
    econ_data = pd.concat([econ_data, monthly], axis=1)

econ_data_q = econ_data.resample('Q').last()
print("Quarterly economic data head:")
print(econ_data_q.head())

# === Step 3: Merge quarterly ETF and economic data ===
data_q = pd.merge(etf_prices_q, econ_data_q, left_index=True, right_index=True, how='inner')

print(f"Merged quarterly data shape: {data_q.shape}")
print("Columns in merged quarterly data:")
print(data_q.columns.tolist())
print(data_q.head())

# === Step 4: Calculate quarterly ETF returns ===
etf_cols = list(etf_tickers.keys())
etf_returns_q = data_q[etf_cols].pct_change().dropna()
print("Quarterly ETF returns head:")
print(etf_returns_q.head())

# Rename ETF columns to sector names
etf_returns_q.rename(columns=etf_tickers, inplace=True)

# === Step 5: Align economic indicators ===
econ_cols = list(fred_series.values())
econ_aligned_q = data_q.loc[etf_returns_q.index, econ_cols]
print("Quarterly economic indicators aligned head:")
print(econ_aligned_q.head())

# === Step 6–7: Combine and compute correlation ===
combined_q = pd.concat([etf_returns_q, econ_aligned_q], axis=1)
corr_matrix_q = combined_q.corr()
corr_etf_econ_q = corr_matrix_q.loc[etf_tickers.values(), fred_series.values()]

print("Correlation matrix (ETF quarterly returns vs Economic indicators):")
print(corr_etf_econ_q)

# === Step 8: Plot heatmap ===
plt.figure(figsize=(12, 8))
sns.heatmap(corr_etf_econ_q, annot=True, cmap="RdYlGn", center=0, fmt=".2f")
plt.title("Correlation: ETF Quarterly Returns vs Economic Indicators (Past 5 Years)")
plt.ylabel("ETF Sectors")
plt.xlabel("Economic Indicators")
plt.tight_layout()
plt.show()
```
